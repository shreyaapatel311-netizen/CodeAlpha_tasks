```python
import matplotlib.pyplot as plt

plt.figure()
plt.hist(df['Age'])
plt.title("Age Distribution of Customers")
plt.xlabel("Age")
plt.ylabel("Count")
mean_age = df['Age'].mean()
plt.axvline(mean_age, linestyle='dashed')
plt.show()
```

### 📊 Insight: Age Distribution of Customers

- The age of customers ranges from 18 to 70.
- The distribution is fairly uniform, with no strong peak.
- Slightly higher concentration is observed around the 40–50 age group.
- The dashed vertical line represents the mean age, which lies near the center.
- This indicates that the average customer is middle-aged.
- Overall, the business serves a wide and diverse age group.

```python
import matplotlib.pyplot as plt

category_sales = df.groupby('Category')['Purchase Amount (USD)'].sum()

plt.figure()
category_sales.plot(kind='bar')
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.show()
```

### 📊 Insight: Total Sales by Category

- Clothing has the highest total sales among all categories.
- Accessories also contribute significantly but are lower than Clothing.
- Footwear shows moderate sales compared to other categories.
- Outerwear has the lowest total sales.
- This indicates that Clothing is the most popular category, while Outerwear has the least demand.
- Overall, sales are concentrated more in Clothing and Accessories categories.

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure()
sns.boxplot(x=df['Purchase Amount (USD)'])
plt.title("Purchase Amount Distribution (Outliers Detection)")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure()
corr = df[['Age', 'Previous Purchases', 'Purchase Amount (USD)', 'Review Rating']].corr()

sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

📊 Correlation Heatmap Insights

- All correlations are very close to 0, indicating no strong relationships between variables  
- Age vs Purchase Amount: ~ -0.01 → no impact  
- Previous Purchases vs Purchase Amount: ~ 0.008 → negligible relationship  
- Purchase Amount vs Review Rating: ~ 0.03 → very weak positive  
- Age vs Review Rating: ~ -0.02 → no meaningful link  

👉 Insight: Customer behavior variables are largely independent , meaning factors like age, past purchases, and ratings do not strongly influence purchase amount . More features may be needed for better predictive analysis.
