```python
df.drop_duplicates(inplace=True)
print("After:", df.shape)
```

```python
df.groupby('Category')['Sales'].sum().plot(kind='bar')
plt.title("Sales by Category")
plt.ylabel("Sales")
plt.show()
```

📊 Sales by Category (Bar Chart)

Insight:
Sales are highest in one specific category, showing that customers prefer that category more than others. This indicates strong market demand and potential focus area for business growth.

```python
df.groupby('Category')['Profit'].sum().plot(kind='bar', color='green')
plt.title("Profit by Category")
plt.show()
```

📊 Profit by Category (Bar Chart)

Insight:
Profit is not equally distributed across all categories. Some categories generate high sales but relatively lower profit, indicating the need to improve cost efficiency or pricing strategy.

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])

df.groupby(df['Order Date'].dt.year)['Sales'].sum().plot(marker='o')
plt.title("Yearly Sales Trend")
plt.show()
```

📊Sales Over Time (Line Chart)

Insight:
Sales show a general increasing trend over time, indicating steady business growth. However, there are some fluctuations, which may be due to seasonal demand or market conditions.

```python
sns.barplot(x='Region', y='Sales', data=df)
plt.title("Sales by Region")
plt.show()
```

📊Region-wise Sales (Bar Chart)

Insight:
Certain regions contribute significantly more to total sales compared to others. This suggests that targeted marketing strategies can be applied to improve performance in low-performing regions.

```python
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

📊Correlation Heatmap

Insight:
The heatmap shows relationships between numerical variables such as sales and profit. A moderate positive correlation indicates that higher sales generally lead to higher profit, but not always proportionally.
